```
from IPython.display import display

print("10 records of dataset")
display(df.head(10))
print("Dataset shape:", df.shape)
print("column names :", df.columns.tolist())
```

```
print('dateset information:')
print("\\n Misssing values in each column :")
display(df.isnull().sum())
print("\\n Statical sumary")
display(df.describe())
```

```
if "student_id" not in df.columns:
    df.insert(0, "student_id", range(1, len(df) + 1))

subjects = ["math score", "reading score", "writing score"]
df["Total_Mark"] = df[subjects].sum(axis=1)
df['average_Marks'] = df[subjects].mean(axis=1)

display(df[["student_id"] + subjects + ["Total_Mark", "average_Marks"]].head(10))
```

```
marks = df[subjects].to_numpy()
print("Maximum score:", np.max(marks))
print("Minimum score:", np.min(marks))
print("Mean score:", round(np.mean(marks), 2))
print("Standard Deviation:", round(np.std(marks), 2))
```

```
top_10 = df.nlargest(10, "average_Marks")
display(top_10[["student_id", "gender"] + subjects + ["average_Marks"]])
```
